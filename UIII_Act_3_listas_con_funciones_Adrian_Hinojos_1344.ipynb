```python
# Función para calcular costo de consulta
def consulta():
    costo = 200
    return costo

# Función para calcular costo de vacuna
def vacuna():
    costo_base = 150
    domicilio = input("¿Es a domicilio? (s/n): ")

    if domicilio.lower() == "s":
        costo_base += 50

    return costo_base

# Función para calcular costo de cirugía
def cirugia():
    costo_base = 500
    peso = float(input("Ingresa el peso del animal (kg): "))

    costo_total = costo_base + (peso * 20)
    return costo_total

# Función menú
def menu():
    opcion = 0

    while opcion != 4:
        print("\n--- VETERINARIA ---")
        print("1. Costo de consulta")
        print("2. Costo de vacuna")
        print("3. Costo de cirugía")
        print("4. Salir")

        opcion = int(input("Elige una opción: "))

        if opcion == 1:
            print("Costo de consulta: $", consulta())

        elif opcion == 2:
            print("Costo de vacuna: $", vacuna())

        elif opcion == 3:
            print("Costo de cirugía: $", cirugia())

        elif opcion == 4:
            print("Saliendo del sistema veterinario...")

        else:
            print("Opción no válida, intenta de nuevo.")

# Llamada al menú
menu()
```
